```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# 데이터 로드
df = pd.read_csv("/Users/sunghyunkim/Desktop/data-QI/autos_onehot.csv")  # CSV 파일로 저장된 경우
X = df.drop(columns=["price"])  # 입력 변수
y = df["price"]  # 타겟 변수
```

```python
df
```

```python
# 데이터 정규화
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.values.reshape(-1, 1)).flatten()  # 1차원 배열로 변환
```

```python
# 데이터 분할 (8:2)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)
```

```python
# MLP 모델 생성
mlp = MLPRegressor(hidden_layer_sizes=(128, 64, 32), activation='relu', solver='adam', max_iter=500)
```

```python
# 모델 학습
mlp.fit(X_train, y_train)
```

```python
# 예측 및 평가
y_pred_scaled = mlp.predict(X_test)
y_pred = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()  # 정규화 복원
```

```python
# 성능 지표 계산
mae = mean_absolute_error(y_test, y_pred_scaled)
mse = mean_squared_error(y_test, y_pred_scaled)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred_scaled)
```

```python

print(f"R^2: {r2}")
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"Root Mean Squared Error: {rmse}")
```
